```python
import re
import string
with open("/content/drive/MyDrive/CNLP_LAB/exp-2/2.1_text_data.txt", "r") as file:
    text = file.read()

print("ORIGINAL TEXT:\n")
print(text)
```

**EXP 2.1**

*BASIC TEXT PREPROCESSING*

```python
# 1. Count uppercase letters
uppercase_count = sum(1 for ch in text if ch.isupper())

# Convert text to lowercase
text = text.lower()

# 2. Count punctuation
punctuation_chars = [ch for ch in text if ch in string.punctuation]
punctuation_count = len(punctuation_chars)
different_punctuations = set(punctuation_chars)

# Remove punctuation
text = text.translate(str.maketrans("", "", string.punctuation))

# 3. Count numbers/digits
number_count = sum(1 for ch in text if ch.isdigit())

# Remove numbers
text = re.sub(r'\d+', '', text)

# 4. Count extra whitespace
# Count whitespace sequences containing 2 or more spaces/newlines/tabs
extra_whitespace = sum(
    len(match.group()) - 1
    for match in re.finditer(r'[ \t]+', text)
)

# Remove extra whitespace
text = re.sub(r'\s+', ' ', text).strip()

# Display results
print("\n--- PREPROCESSING INFORMATION ---")
print("Number of uppercase letters:", uppercase_count)
print("Total punctuation marks:", punctuation_count)
print("Different types of punctuation:", different_punctuations)
print("Number of digits:", number_count)
print("Extra whitespace characters:", extra_whitespace)

print("\n--- CLEANED TEXT ---")
print(text)
```

**EXP 2.2**

*WORD AND SENTENCE TOKENIZATION*

```python
with open("/content/drive/MyDrive/CNLP_LAB/exp-2/2.2_tokenization_data.txt", "r") as file:
    text = file.read()

import nltk

nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize, sent_tokenize
with open("/content/drive/MyDrive/CNLP_LAB/exp-2/2.2_tokenization_data.txt", "r") as file:
    text = file.read()

# Sentence tokenization
sentences = sent_tokenize(text)

# Word tokenization
words = word_tokenize(text)

print("----- SENTENCE TOKENS -----")

for i, sentence in enumerate(sentences, 1):
    print(i, ":", sentence)

print("\n----- WORD TOKENS -----")

for i, word in enumerate(words, 1):
    print(i, ":", word)
```

```python
import spacy

# Load English language model
nlp = spacy.load("en_core_web_sm")

# Process text
doc = nlp(text)

# Sentence tokenization
print("----- SENTENCE TOKENS -----")

for i, sentence in enumerate(doc.sents, 1):
    print(i, ":", sentence.text)

# Word tokenization
print("\n----- WORD TOKENS -----")

for i, token in enumerate(doc, 1):
    print(i, ":", token.text)
```

```python
import re


# Sentence tokenization
sentences = re.split(r'(?<=[.!?])\s+', text.strip())

print("----- SENTENCE TOKENS -----")

for i, sentence in enumerate(sentences, 1):
    print(i, ":", sentence)

# Word tokenization
words = re.findall(r"\b\w+\b|[^\w\s]", text)

print("\n----- WORD TOKENS -----")

for i, word in enumerate(words, 1):
    print(i, ":", word)
```
