```python
from nltk.stem import PorterStemmer

# Create Porter Stemmer object
ps = PorterStemmer()

# Input words
words = [
    "playing",
    "played",
    "plays",
    "studies",
    "studying",
    "connected",
    "connection",
    "computers"
]

# Apply stemming
for word in words:
    stem = ps.stem(word)
    print(word, "->", stem)
```

```python
import re

text = """
Welcome to the Natural Language Processing workshop!
For more information, visit https://www.nlpworkshop.com or www.python.org.
You can contact the coordinator at nlpworkshop@gmail.com or support@python.org.
For registration, call +91-9876543210 or 9123456789.
Follow us on social media @NLPWorkshop and @PythonLearner.
Share your experience using #NLP, #Python, and #MachineLearning.
You can also visit https://github.com/NLPWorkshop for the latest updates.
"""

# 1. Extract Email Addresses
emails = re.findall(
    r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}',
    text
)

# 2. Extract URLs
urls = re.findall(
    r'https?://[^\s]+|www\.[^\s]+',
    text
)

# 3. Extract Mobile Numbers
mobile_numbers = re.findall(
    r'(?:\+91[-\s]?)?[6-9]\d{9}',
    text
)

# 4. Extract Hashtags
hashtags = re.findall(
    r'#\w+',
    text
)

# 5. Extract Mentions
mentions = re.findall(
    r'@\w+',
    text
)

print("Email Addresses:")
print(emails)

print("\nURLs:")
print(urls)

print("\nMobile Numbers:")
print(mobile_numbers)

print("\nHashtags:")
print(hashtags)

print("\nMentions:")
print(mentions)
```
