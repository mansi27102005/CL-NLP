```python
import pandas as pd
df=pd.read_csv('employee_information_100.csv')
df.head()
```

```python
import matplotlib.pyplot as plt
print(df.columns)
```

**EXP 1.1**

Read the employee dataset and display the average salary of each department using a bar chart.


```python
department="Department"
salary="Salary"

avg_salary = df.groupby(department)[salary].mean()

print(avg_salary)
print()

plt.figure(figsize=(8,5))
avg_salary.plot(kind='bar',color='pink')
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.show()
```

**EXP 1.2**

Display the number of employees working in each department.



```python
employee_id="Employee_ID"
employee_count=df.groupby(department)[employee_id].count()

print(employee_count)
```

**EXP 1.3**

Display the percentage of male and female employees using a pie chart


```python
gender="Gender"
gender_count = df[gender].value_counts()

plt.figure(figsize=(6,6))
plt.pie(gender_count,
        labels=gender_count.index,
        autopct="%1.1f%%",
        startangle=90)
plt.title("Gender Distribution")
plt.show()
```

**EXP 1.4**

Draw a histogram showing salary distribution.

```python
plt.figure(figsize=(8,5))
plt.hist(df[salary], bins=10,color='purple')
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```

**EXP 1.5**

Draw a scatter plot showing the relationship between experience and salary.


```python
experience="Experience_Years"

plt.figure(figsize=(8,5))
plt.scatter(df[experience], df[salary],color='red')
plt.title("Experience vs Salary")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.show()
```

**EXP 1.6**

Display the top 10 employees with the highest salary

```python
print("\nTop 10 Highest Paid Employees:")
print(df.nlargest(10, salary)[[employee_id, department, salary]])
```

**EXP 1.7**

Find the highest salary in every department.


```python
print("\nHighest Salary in Every Department:")
print(df.groupby(department)[salary].max())
```

**EXP 1.8**

Display employees whose salary is greater than the overall average salary

```python
overall_avg = df[salary].mean()

print("\nEmployees with Salary Greater Than Overall Average:")
print(df[df[salary] > overall_avg])
```

**EXP 1.9**

Calculate the average years of experience in each department

```python
print("\nAverage Experience by Department:")
print(df.groupby(department)[experience].mean())
```

**EXP 1.10**

Display the distribution of employee ages using a histogram.


```python
age="Age"
plt.figure(figsize=(8,5))
plt.hist(df[age], bins=10,color='teal')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```
